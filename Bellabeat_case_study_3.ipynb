**Identify trends in tracker app usage**

what is the average time spent with the tracker on a day?


```python
#get the total minutes active
activity['TotalMinutes'] = activity['FairlyActiveMinutes'] + activity['LightlyActiveMinutes'] + activity['VeryActiveMinutes'] + activity['SedentaryMinutes']

#convert the total minutes to hours
activity['Active_Hours'] = activity['TotalMinutes']//60

activity.head()
```

```python
activity['Active_Hours'].describe()
```

On Average, the users have the tracker active for 20 hours a day.

```python
#what is the trends in the tracker usage based on the day of the week
activity['day'] = activity['date'].dt.day_name()
activity.head()
```

```python
weekly_average = activity.groupby('day')['Active_Hours'].mean()
order = ['Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday']
weekly_average = weekly_average.reindex(order)
weekly_average.head(7)
```

```python
#Plot a graph of the average time the tracker is active on each day of the week

plt.figure(figsize=(7,4))
sns.barplot(
            x= weekly_average.index,
            y= weekly_average.values

           )
sns.color_palette()
plt.title('The average time the tracker is active on given day of the week',fontsize='small')
plt.ylabel('Average hours',fontsize='small')
plt.xlabel('Day of the week')
plt.xticks(rotation = 45)
plt.show()
```

There is no noticeable difference in the average number of hours the tracker is used on any specific day of the week. This means the users consistently use/wear the tracker.

**Catergorize the users**

 We will use the average number of total steps taken in a day to catergorize the users.  We will catergorize them as sedentary, Active, and Highly Active.

```python
average_daily_steps = activity.groupby('Id')['TotalSteps'].mean().reset_index().round(0)
average_daily_steps.head()
```

```python
average_daily_steps['category'] = pd.cut(average_daily_steps['TotalSteps'],
                                         bins=[-np.inf, 5000, 7500, 10000, 12500, np.inf],
                                         labels=["Sedentary","Low Active","Somewhat Active","Active","Highly Active"],
                                         right=False)
average_daily_steps.head()
```

```python
propotion_activity = round(average_daily_steps['category'].value_counts(normalize=True)*100)
propotion_activity.head()
```

```python
plt.figure(figsize=(7,4))
plt.pie(x = propotion_activity.values,
        labels = propotion_activity.index,
        autopct="%1.1f%%")
plt.title('The proportion of users in each category of physical activity', fontsize ='small')
plt.xlabel('Category',fontsize ='small')
plt.show()
```

37% of the users live a sedentary lifestyle, taking less than 5000 steps a day. Only 6% of the users are highly active, taking more than 12500 steps per day.

```python
#Average steps on each day of the week
average_daily_steps = activity.groupby('day')['TotalSteps'].mean().round()
average_daily_steps = average_daily_steps.reindex(order)

average_daily_steps.head(7)
```

```python

plt.figure(figsize=(7,4))
ax = sns.barplot(x=average_daily_steps.index,
            y= average_daily_steps.values,
                 hue = average_daily_steps.index,
            palette=sns.color_palette("viridis", n_colors=len(average_daily_steps)),
    legend=False
                 )
plt.axhline(7000,color='red')
ax.text(9,7000,"Daily recommended steps",va='top',ha='right',color='green')
plt.title('Average steps taken on each day of the weak',fontsize='small')
plt.xlabel('Day of the week', fontsize='small')
plt.ylabel('Average steps taken')
plt.xticks(rotation =45)
plt.show()
```

Users are reaching the recommended daily steps on all days except sunday. It seems users spend sunday at home resting, which may explain why they don't reach the daily recommended steps.

Users can be sent a notification on sundays to nudge them to take extra steps to reach the daily recommended goal.

```python
#calculate  the time individuals spend in bed  before falling  sleep.
sleep['time_in_bed_before_sleep'] = sleep['TotalTimeInBed'] - sleep['TotalMinutesAsleep']
```

```python
#convert sleepday to datetime and extract the day of the week
sleep['date'] = pd.to_datetime(sleep['SleepDay'])
sleep['day'] = sleep['date'].dt.day_name()
sleep.head()
```

```python
#convert sleep minutes to hours
sleep['hours'] = sleep['TotalMinutesAsleep']//60
sleep.head()
```

```python
average_daily_sleep = sleep.groupby('day')[['TotalMinutesAsleep','time_in_bed_before_sleep']].mean().round()
average_daily_sleep = average_daily_sleep.reindex(order)
average_daily_sleep.head()
```

```python
average_daily_sleep.describe()
```

```python
average_without_sleep = round(average_daily_sleep['time_in_bed_before_sleep'].mean())
print(average_without_sleep)
```

```python
plt.figure(figsize=(7,4))
ax = sns.barplot(x=average_daily_sleep.index,
            y= average_daily_sleep['TotalMinutesAsleep'],
                 palette='Greens',
                 hue=average_daily_sleep.index,
                 legend=False)
plt.axhline(420,color='red')
ax.text(9.5,420,"Recommended minutes of sleep",va='top',ha='right',color='green')
plt.title('Average minutes of sleep on each day of the weak',fontsize='small')
plt.xlabel('Day of the week', fontsize='small')
plt.ylabel('Average minutes of sleep')
plt.xticks(rotation =45)
plt.savefig("average_minutes of sleep.png")
plt.show()
```

```python
# correlation between sleep and calories
#we need to firs merge the sleep and activity dataset using ID and date as the common Keys between the two datasets.
activity_sleep_df = pd.merge(activity,sleep, on=['Id','date'],how='inner')
```

```python
activity_sleep_df.head()
```

```python
plt.figure(figsize=(7,4))
sns.regplot(x=activity_sleep_df['Calories'],
            y=activity_sleep_df['TotalMinutesAsleep'],
            line_kws={"color": "red"},
            color='green')
plt.title('The relationship between Sleep and calories burned',fontsize='medium')
plt.xlabel('Calories')
plt.ylabel('Total minutes a sleep')
plt.show()
```

```python
plt.figure(figsize=(7,4))
sns.regplot(y=activity_sleep_df['Calories'],x=activity_sleep_df['TotalSteps'],line_kws={"color": "red"},color='green')
plt.title('The relationship between Sleep and calories burned',fontsize='medium')
plt.xlabel('steps')
plt.ylabel('Calories ')
plt.show()
```

There is a postive relationship between steps taken and calories expended. Meaning the more steps an the user takes the more calories they expend.

There is no relationship between time asleep and calories expended. This is intuitive because during sleep the body isn't not burning much calories to find a relationship.

**KEY FINDINGS**


*   On average users wear the tracker for **20 hours a day**. Meaning the tracker is active and able to collect user data for most of the day.
*   The users wear the tracker consistently. There is no preference on the day of the week when the users wear the tracker.
*   Majority (37%)of the users live a sedentary life style, taking less than 5000 average steps a day. Only 6% of users are highly active, taking more than 12500 steps a day.
*  On average, users make 7000 steps a day, except on Sundays.(This average daily steps could be skewed by those taking more steps)
* Users are most active between 9am and 8pm, taking the most steps during this time. The steps taken peak at noon, and between 6-7pm.
* Users sleep less than the recommended time(atleast 7 hours of sleep), except on Sunday and wednesday where they  hit more than 7 hours of sleep. They spend an average of 40 minutes in bed without sleeping.
* There is relationship betweeen steps taken and calories expended.The more steps taken, the more calories expended.
