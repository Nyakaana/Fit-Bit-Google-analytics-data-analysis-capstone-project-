#**ANALYSIS**

##**Identify trends in tracker app usage**




###**1) what is the average time the tracker is active per day?**

```python
#get the total minutes active per day
activity['TotalMinutes'] = activity['FairlyActiveMinutes'] + activity['LightlyActiveMinutes'] + activity['VeryActiveMinutes'] + activity['SedentaryMinutes']

#convert the total minutes to hours
activity['Active_Hours'] = activity['TotalMinutes']//60

# calculate the average time .
hours_active_per_day = round(activity['Active_Hours'].mean(),0)

print(f"The tracker is active for {hours_active_per_day} hours on average per day")
```

###**2) what are the trends in the tracker usage based on the day of the week**


```python
# Extract the day of the week from the date column
activity['day'] = activity['date'].dt.day_name()
activity.head()
```

```python
# Average hours the tracker is active on a specific day of the week
weekly_average = round(activity.groupby('day')['Active_Hours'].mean(),0)
order = ['Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday']
weekly_average = weekly_average.reindex(order)
weekly_average.head(7)
```

```python
# Plot a graph of the average time the tracker is active on each day of the week

plt.figure(figsize=(7,4))
ax= sns.barplot(
            x= weekly_average.index,
            y= weekly_average.values,
            palette="YlGn",
            hue = weekly_average.values,
            edgecolor="gray",
            legend = False

           )

plt.title('The average time the tracker is active on a specific day of the week',fontsize='medium')
plt.ylabel('Average hours',fontsize='small')
plt.xlabel('Day of the week')
plt.axhline(hours_active_per_day,color='green')
ax.text(9,hours_active_per_day,"Overall average hours daily\n the tracker is active",va='top',ha='right',color='blue')
plt.xticks(rotation = 45)
plt.show()
```

**Insights from the analysis above**

* There is no significant difference in the average number of hours the tracker is used on any specific day of the week. This means the users consistently use/wear the tracker.The day of the week doesn't seem to determine the usage of the tracker.

##**Catergorize the users based on their daily average steps taken**

 We will use the average number of total steps taken in a day to catergorize the users.  We will catergorize them as sedentary, Active, and Highly Active.

```python
# Get the average daily steps taken by each user
average_daily_steps = activity.groupby('Id')['TotalSteps'].mean().reset_index().round(0)
average_daily_steps.head()
```

```python
# Create categories to group each user based on their average steps taken per day
average_daily_steps['category'] = pd.cut(average_daily_steps['TotalSteps'],
                                         bins=[-np.inf, 5000, 7500, 10000, 12500, np.inf],
                                         labels=["Sedentary","Low Active","Somewhat Active","Active","Highly Active"],
                                         right=False)
average_daily_steps.head()
```

```python
# Find the proportion of users in each of the categories created above
propotion_activity = round(average_daily_steps['category'].value_counts(normalize=True)*100)
propotion_activity.head()
```

```python
plt.figure(figsize=(7,4))
plt.pie(x = propotion_activity.values,
        labels = propotion_activity.index,
        colors = sns.color_palette("YlGn"),
        autopct="%1.1f%%"
        )
plt.title('The proportion of users in each category of daily physical activity', fontsize ='medium')
plt.xlabel('Category',fontsize ='small')
plt.show()
```

**Insights from the analysis above**

* 37% of the users live a sedentary lifestyle, taking less than 5000 steps daily. Only 6% of the users are highly active, taking more than 12500 steps daily.

* 57 % of the users are living a low activity and sedementary lifestyle.




##**Explore patterns in the steps taken on particular days of the week**

```python
# Average steps on each day of the week
average_daily_steps = activity.groupby('day')['TotalSteps'].mean().round()
average_daily_steps = average_daily_steps.reindex(order)

average_daily_steps.head(7)
```

**Insights from the analysis**


* On Average Users are reaching the recommended daily steps on all days except sunday. This suggests that  users spend sunday at home resting, which may explain why they don't reach the daily recommended steps.

* Earlier analysis, showed that most users are in the sedentary and low activity category and this analysis shows that on average the Users hit the recommended number of steps a day. These two findings contradict each other.  

* Segmentation analysis should be conducted to analyze the behavior of users in different categories. This will help us determine whether the daily average steps per user is skewed by a few active and highly active users taking significantly more steps daily.

##**Explore when users are most and least active (steps taken) during the day**

* This will give us insights on when to send nudges to encourage users to take steps.

```python
#calculate  the time individuals spend in bed  before falling  sleep.
sleep['time_in_bed_before_sleep'] = sleep['TotalTimeInBed'] - sleep['TotalMinutesAsleep']
```

```python
#convert sleep minutes to hours
sleep['hours'] = sleep['TotalMinutesAsleep']//60
sleep.head()
```

```python
# calculate the average minutes of sleep
average_daily_sleep = sleep.groupby('day')[['TotalMinutesAsleep','time_in_bed_before_sleep']].mean().round()
average_daily_sleep = average_daily_sleep.reindex(order)
average_daily_sleep.head()
```

```python
# Average minutes users spend in bed before sleeping
average_without_sleep = round(average_daily_sleep['time_in_bed_before_sleep'].mean())
print(f"Users spend on average {average_without_sleep} minutes in bed before sleeping")
```

```python
# Plot of Average daily  minutes of sleep
plt.figure(figsize=(7,4))
ax = sns.barplot(x=average_daily_sleep.index,
            y= average_daily_sleep['TotalMinutesAsleep'],
                 palette="YlGn",
                 hue=average_daily_sleep['TotalMinutesAsleep'],
                 legend=False,
                 edgecolor = "gray")
plt.axhline(420,color='green')
ax.text(9,420,"Recommended minutes of sleep",va='top',ha='right',color='blue',fontsize ='small')
plt.title('Average minutes of sleep on each day of the weak',fontsize='medium')
plt.xlabel('Day of the week', fontsize='small')
plt.ylabel('Average minutes of sleep')
plt.xticks(rotation =45)
plt.savefig("average_minutes of sleep.png")
plt.show()
```

```python
# Graph of the total time in bed before user sleep

plt.figure(figsize=(7,4))
ax = sns.barplot(x=average_daily_sleep.index,
            y= average_daily_sleep['time_in_bed_before_sleep'],
                 palette=sns.color_palette("YlGn_r"),
                 hue=average_daily_sleep['time_in_bed_before_sleep'],
                 legend=False,
                 edgecolor ="gray")
plt.axhline(average_without_sleep,color='green')
plt.text(6.6,average_without_sleep,'Overall average time in bed without sleep',fontsize='small',color='blue')
plt.title('Average minutes in bed  without sleep on each day of the weak',fontsize='small')
plt.xlabel('Day of the week', fontsize='small')
plt.ylabel('Average minutes of time without sleep',fontsize='small')
plt.xticks(rotation =45)
plt.show()
```

##**Correlation between sleep with steps taken and calories burned**


```python
# correlation between sleep and calories
#we need to firs merge the sleep and activity dataset using ID and date as the common Keys between the two datasets.
activity_sleep_df = pd.merge(activity,sleep, on=['Id','date'],how='inner')
```

```python
activity_sleep_df.head()
```

```python
# The relationship between sleep and Calories burned
plt.figure(figsize=(7,4))
sns.regplot(x=activity_sleep_df['Calories'],
            y=activity_sleep_df['TotalMinutesAsleep'],
            line_kws={"color": "red"},
            color='green')
plt.title('The relationship between Sleep and calories burned',fontsize='medium')
plt.xlabel('Calories')
plt.ylabel('Total minutes a sleep')
plt.show()
```

```python
# The relationship between sleep and Calories burned
plt.figure(figsize=(7,4))
sns.regplot(y=activity_sleep_df['Calories'],x=activity_sleep_df['TotalSteps'],line_kws={"color": "red"},color='green')
plt.title('The relationship between Sleep and total steps taken',fontsize='medium')
plt.xlabel('steps')
plt.ylabel('Calories ')
plt.show()
```

**Insights from the analysis**

* Users sleep less than the recommended hours of sleep except on Sunday and wednesday.

* It is only on sunday and Wednesday that the users have sufficient sleep. With the most sleep experienced on sunday.

* Also, they spend an average of 40 minutes in bed without sleeping. On sundays. they spend more than the average time in bed without sleeping .

* There is a positive relationship between steps taken and calories expended. Meaning the more steps an the user takes the more calories they expend.

* There is no relationship between time asleep and calories expended. This is intuitive because during sleep the body isn't not burning calories to form a  relationship.
